#Descripción del modelo computacional

El artículo original analiza el flujo de producción y la confiabilidad de una línea de prensas mecánicas perteneciente a la industria automotriz. Físicamente, el sistema de manufactura está constituido por una estación de entrada, 5 máquinas prensadoras conectadas en serie, una estación de salida y 6 puestos de manipulación encargados de la carga, descarga y transferencia de piezas entre etapas sucesivas. Se plantean tres modelos: ideal (modelo 1), operado por robots (modelo 2) y operado por humanos (modelos 3).



## Modelo ideal (1)

Está compuesto únicamente por las 5 máquinas operando al 100% de utilización sin fallas ni demoras de transferencia, lo usamos para determinar la capacidad máxima teórica de producción ($MaxLimit$).

### Funciones
La cantidad de productos producida es:
\begin{align}
P(t) &= \frac{t_{\text{produccion}}(t)}{t_{cicloMaquina}} \\[4pt]
\end{align}

##Modelo operado por robots (2)

Incorpora robots que operan a velocidad constante y sincronizados, contemplando pausas planificadas por tiempos de preparación, y pausas no planificadas por fallas aleatorias de máquinas y robots en la producción.


###Distribuciones
Las variables siguen las siguientes distribuciones.
Siendo $N_R(t)$ la cantidad de fallas producidas por un robot en el tiempo $t$, $N_M(t)$ la cantidad de fallas producidas por una máquina en el tiempo $t$ y $c_i$ la probabilidad de que el producto sea de buena calidad.

$$\begin{align}
N_R(t) &\sim \text{Poisson}\!\left(\frac{t}{\text{MTBF}_{R}^{\text{total}}}\right) \\[4pt]
N_M(t) &\sim \text{Poisson}\!\left(\frac{t}{\text{MTBF}_{M}^{\text{total}}}\right) \\[4pt]
c_i &\sim U(0,1)
\end{align}$$

### Funciones
Considerando que se tienen m máquinas y r robots, el tiempo medio de reparacion entre máquinas ($MTBF_{M}^{total}$) y entre robots ($MTBF_{R}^{total}$), se calcula como:

\begin{align}
\left(\text{MTBF}_{R}^{\text{total}}\right)^{-1} &= \sum_{i=1}^{r} \left(\text{MTBF}_{R,i}\right)^{-1} \\[4pt]
\left(\text{MTBF}_{M}^{\text{total}}\right)^{-1} &= \sum_{i=1}^{m} \left(\text{MTBF}_{M,i}\right)^{-1} \\[4pt]
\end{align}
El tiempo de producción total se calcula como:
\begin{align}
t_{\text{produccion}}(t) &= t
  - \left\lfloor \frac{t}{t_{turno}} \right\rfloor t_{\text{preparacion}}
  - t_{\text{reparacion},R}\, N_R(t)
  - t_{\text{reparacion},M}\, N_M(t) \\[4pt]
\end{align}
A partir de eso, se llega a que la cantidad de productos producida es:
\begin{align}
P(t) &= \frac{t_{\text{produccion}}(t)}{t_{cicloMaquina}+t_{movimientoRobot}} \\[4pt]
\end{align}

Donde, para cada producto se calcula su estado considerando su calidad ($c_i$), siendo 1 si es de buena calidad y 0 si se considera defectuoso:
\begin{align}
\text{estado}_i &=
\begin{cases}
1 & \text{si } c_i \leq P_{b,r} \\
0 & \text{si } c_i > P_{b,r}
\end{cases} \\[4pt]
\end{align}
Donde $P_{b,r}$ es la probabilidad de ser defectuoso dado que la linea de producción esta contituida por robots.

Obteniendo entonces la produccion de productos de buena calidad ($Pb(t)$) y la cantidad de defectuosos ($Pd(t)$):
\begin{align}
P_b(t) &= \sum_{i=1}^{P(t)} \text{estado}_i \\[4pt]
P_d(t) &= P(t) - P_b(t)
\end{align}

### Parametros


\begin{aligned}
t_m &= 5 \text{ s} & t_r &= 1 \text{ s} \\
t_{\text{preparación}} &= 900 \text{ s} & t_{\text{turno}} &= 28800 \text{ s}  \\
\\
MTBF_m^{(1)} &= 1800000 \text{ s} & t_{reparacion,M} &= 14400 \text{ s} \\
\\
MTBF_r^{(1)} &= 3600000 \text{ s} & MTBF_r^{(2)} &= 7200000 \text{ s} \\
t_{reparacion,R} &= 14400 \text{ s} \\
\\
 P_{br} &= 0.99999
\\
m &= 5 & r &= 6
\end{aligned}


## Modelo operado por Humanos (3)

Incorpora trabajadores humanos que operan con tiempos de manipulación con variabilidad estocástica, contemplando pausas planificadas de descanso y preparación y pausas no planificadas por fallos aleatorios de las maquinas o errores humanos en la produccion.


###Distribuciones
Las variables siguen las siguientes distribuciones.
Siendo $N_H(t)$ la cantidad de errores humanos producidos por los operadores en el tiempo $t$, $N_M(t)$ la cantidad de fallas producidas por las máquinas en el tiempo $t$, $c_i$ la calidad del producto $i$ y $t_o$ el tiempo que tarda el operador en cargar o descargar una máquina.

\begin{align}
N_H(t) &\sim \text{Poisson}\!\left(\frac{t}{\text{MTBF}_{H}^{\text{total}}}\right) \\[4pt]
N_M(t) &\sim \text{Poisson}\!\left(\frac{t}{\text{MTBF}_{M}^{\text{total}}}\right) \\[4pt]
t_o &\sim N(10,\,2) \\[4pt]
c_i &\sim U(0,1)
\end{align}



### Funciones
Considerando que se tienen $m$ máquinas y $h$ operadores, el tiempo medio entre fallas de las máquinas ($\text{MTBF}_{M}^{\text{total}}$) y entre errores de los operadores ($\text{MTBF}_{H}^{\text{total}}$) se calcula como:

\begin{align}
\left(\text{MTBF}_{H}^{\text{total}}\right)^{-1} &= \sum_{i=1}^{h} \left(\text{MTBF}_{H,i}\right)^{-1} \\[4pt]
\left(\text{MTBF}_{M}^{\text{total}}\right)^{-1} &= \sum_{i=1}^{m} \left(\text{MTBF}_{M,i}\right)^{-1}
\end{align}

El tiempo de producción total se calcula como:
\begin{align}
t_{\text{produccion}}(t) &= t
  - \left\lfloor \frac{t}{t_{turno}} \right\rfloor \left(t_{\text{preparacion}} + t_{\text{descanso}}\right)
  - t_{\text{reparacion},M}\, N_M(t)
  - t_{\text{correccion}}\, N_H(t)
\end{align}

A partir de eso, se llega a que la cantidad de productos producida es:
\begin{align}
P(t) &= \frac{t_{\text{produccion}}(t)}{t_{\text{cicloMaquina}} + t_{o}}
\end{align}

Donde, para cada producto se calcula su estado considerando su calidad ($c_i$), siendo 1 si es de buena calidad y 0 si se considera defectuoso:
\begin{align}
\text{estado}_i &=
\begin{cases}
1 & \text{si } c_i \leq P_{b,h} \\
0 & \text{si } c_i > P_{b,h}
\end{cases}
\end{align}
Donde $P_{b,h}$ es la probabilidad de que un producto sea de buena calidad dado que la línea de producción está constituida por operadores humanos.

Obteniendo entonces la producción de productos de buena calidad ($P_b(t)$) y la cantidad de defectuosos ($P_d(t)$):
\begin{align}
P_b(t) &= \sum_{i=1}^{P(t)} \text{estado}_i \\[4pt]
P_d(t) &= P(t) - P_b(t)
\end{align}

### Parametros


\begin{aligned}
t_m &= 5 \text{ s} & t_r &= 1 \text{ s} \\
t_{\text{preparación}} &= 900 \text{ s} & t_{\text{turno}} &= 28800 \text{ s}  \\
\\
MTBF_m^{(1)} &= 1800000 \text{ s} &
t_{reparacion,M} &= 14400 \text{ s} \\
\\
MTBF_h &= 28800 \text{ s} & t_{correccion} &= 300 \text{ s} \\
\\
Pb_h &= 0.9973
\\
m &= 5 & h &= 6
\end{aligned}


## Métricas: OEE

La eficiencia del uso de los medios de producción se expresa mediante la métrica OEE, que depende dela cantidad de piezas de buena calidad y el total ideal teórico producido en ese tiempo.
\begin{align}
\text{OEE} = \frac{P_b}{\text{maxLimit}}
\end{align}

```python
# Cantidad de máquinas
M = 5

# Horizontes de simulación en horas
tiempos = [8, 24, 2000, 6000]

# Cantidad de repeticiones por horizonte
N = 50

# Semilla para reproducibilidad
SEMILLA = 42

np.random.seed(SEMILLA)
```

```python
resultados_robot = []

for T in tiempos:

    producciones = []
    oees = []

    print(f"Ejecutando simulaciones para T = {T} horas...")

    for i in range(N):

        fallas_r, fallas_m, P, Pb, Pd, oee = m2.Modelo_robots(T, M)

        producciones.append(P)
        oees.append(oee)

    # Estadísticas de producción
    Pavg = np.mean(producciones)
    desviacion = np.std(producciones, ddof=1)

    # Desviación relativa
    delta = desviacion / Pavg if Pavg != 0 else np.nan

    # Límite máximo de producción
    max_limit = (T * p.hora) / p.ciclo_maquina

    # OEE promedio
    oee_promedio = np.mean(oees)

    resultados_robot.append({
        "Tiempo [h]": T,
        "Max Limit [PCs.]": max_limit,
        "Pavg [PCs.]": Pavg,
        "Desviación estándar [PCs.]": desviacion,
        "Delta": delta,
        "OEE": oee_promedio
    })

print("Simulaciones finalizadas.")
```

```python
import numpy as np
import pandas as pd

tiempos = [8, 24, 2000, 6000]

articulo_humano = [
    dict(T=8, ML=5760, P=1681, S=50.72, D=0.030, O=0.2918),
    dict(T=24, ML=17280, P=5111, S=85.28, D=0.0167, O=0.2958),
    dict(T=2000, ML=1440000, P=427179, S=714.96, D=0.0016, O=0.2967),
    dict(T=6000, ML=4320000, P=1281345, S=1207, D=0.0009, O=0.2966)
]

articulo_robot = [
    dict(T=8, ML=5760, P=4404, S=619.3, D=0.1406, O=0.7646),
    dict(T=24, ML=17280, P=13300, S=1432, D=0.1077, O=0.7697),
    dict(T=2000, ML=1440000, P=1094949, S=19331, D=0.0176, O=0.7604),
    dict(T=6000, ML=4320000, P=3279888, S=27496, D=0.0084, O=0.7592)
]

filas = [
    "Max Limit [PCs.]",
    "Pavg [PCs.]",
    "Desviación estándar [PCs.]",
    "Delta",
    "OEE"
]

def formato_resultados(lista):
    return [{
        "Tiempo [h]": r["T"],
        filas[0]: r["ML"],
        filas[1]: r["P"],
        filas[2]: r["S"],
        filas[3]: r["D"],
        filas[4]: r["O"]
    } for r in lista]

res_art_h = formato_resultados(articulo_humano)
res_art_r = formato_resultados(articulo_robot)


def construir_tabla(humanos, robots):
    datos = {}

    for T in tiempos:
        for modelo, resultados in [("Humanos", humanos), ("Robots", robots)]:
            resultado = next(r for r in resultados if r["Tiempo [h]"] == T)
            datos[(f"{T} h", modelo)] = [resultado[f] for f in filas]

    return pd.DataFrame(datos, index=filas)


def mostrar(tabla):
    estilo = tabla.style
    for idx in tabla.index:
        nombre = idx[0] if isinstance(idx, tuple) else idx
        if nombre in ("Max Limit [PCs.]", "Pavg [PCs.]"):
            fmt = "{:,.0f}"
        elif nombre == "Desviación estándar [PCs.]":
            fmt = "{:,.1f}"
        else:
            fmt = "{:.4f}"
        estilo = estilo.format(fmt, subset=pd.IndexSlice[[idx], :])
    display(estilo)


tabla_obtenida = construir_tabla(resultados_humano, resultados_robot)
tabla_articulo = construir_tabla(res_art_h, res_art_r)

tabla_comb = pd.concat(
    [tabla_obtenida, tabla_articulo],
    keys=["Obtenido", "Artículo"]
).swaplevel(0, 1)

orden = pd.MultiIndex.from_product([filas, ["Obtenido", "Artículo"]])
tabla_comb = tabla_comb.loc[orden]


print("Comparación:")
mostrar(tabla_comb)
```
